```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('qoute_dataset.csv')
```

```python
df.shape
```

```python
qouts = df['quote']
```

```python
qouts.head()
```

```python
qouts = qouts.str.lower()
```

```python
import string
translator = str.maketrans('','',string.punctuation)
qouts=qouts.apply(lambda x:x.translate(translator) )
```

```python
qouts.head()
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
vocb_size=9000
tokinzer = Tokenizer(num_words=vocb_size)
tokinzer.fit_on_texts(qouts)
```

```python
word_indx = tokinzer.word_index
print(len(word_indx))
list(word_indx.items())[:10]
```

```python
sequence = tokinzer.texts_to_sequences(qouts)
```

```python
for i in range(3):
  print(qouts[i])
  print("/n")
  print(sequence[i])
```

```python
x=[]
y=[]

for seq in sequence:
  for i in range(len(seq)):
    input_seq=seq[:i]
    out_seq = seq[i]
    x.append(input_seq)
    y.append(out_seq)
```

```python
len(x)
```

```python
maxlen = max(len(n) for n in x)
```

```python
maxlen
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
x_padded = pad_sequences(x,maxlen=maxlen,padding='pre')
```

```python
y=np.array(y)
```

```python
x_padded.shape
```

```python
from tensorflow.keras.utils import to_categorical
y_cat = to_categorical(y,num_classes=vocb_size)
```

```python
y_cat.shape
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,LSTM,Dense,SimpleRNN
```

```python
embeddings=50
rnn_uniits=128
```

```python
rnn_model = Sequential([
    Embedding(vocb_size,embeddings,input_length=maxlen),
    SimpleRNN(rnn_uniits),
    Dense(vocb_size,activation='softmax')


])
```

```python
rnn_model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
rnn_model.summary()
```

```python
lstm_model = Sequential([
    Embedding(vocb_size,embeddings,input_length=maxlen),
    LSTM(rnn_uniits),
    Dense(vocb_size,activation='softmax')
])
```

```python
lstm_model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
lstm_model.summary()
```

```python
lstm_model.fit(x_padded,y_cat,epochs=20,verbose=0)
```
