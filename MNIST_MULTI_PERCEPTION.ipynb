```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms

transform_pipeline =transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307),(0.3081))
])
print("--- System Loader: Fetching MNIST Benchmark Data ---")
train_dataset = torchvision.datasets.MNIST(root='./data',train=True,download=True,transform=transform_pipeline)
test_dataset = torchvision.datasets.MNIST(root='./data',train = False,download=True,transform=transform_pipeline)


train_loader = torch.utils.data.DataLoader(train_dataset,batch_size=64,shuffle=True)
test_loader = torch.utils.data.DataLoader(test_dataset,batch_size=1000,shuffle=False)


class Multilayerperception(nn.Module):
  def __init__(self):
    super(Multilayerperception,self).__init__()
    self.flatten = nn.Flatten()


    self.hidden_layer = nn.Linear(in_features=784, out_features=128)
    self.relu = nn.ReLU()

    self.output_layer=nn.Linear(in_features=128,out_features=10)


  def forward(self,x):
    x = self.flatten(x)
    z1 = self.hidden_layer(x)
    a1 = self.relu(z1)
    logits = self.output_layer(a1)

    return logits
mlp_model = Multilayerperception()

loss_criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(mlp_model.parameters(),lr=0.01)
print("\n--- Starting Mainframe Classification Run ---")

mlp_model.train()
for epoch in range(3):
  running_loss=0.0
  for batch_idx,(images,targets) in enumerate(train_loader):

    output_logits = mlp_model(images)
    loss = loss_criterion(output_logits,targets)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (batch_idx + 1)%300 ==0:
      print(f"Epoch [{epoch+1}/3] | Batch [{batch_idx+1}/{len(train_loader)}] | Step Loss: {loss.item():.4f}")

# Step 5: Run Production Validation Evaluation
mlp_model.eval() # Switch network into evaluation mode
correct_predictions = 0
total_samples = 0

print("\n--- Evaluating Model Accuracy Across Unseen Test Archive ---")
with torch.no_grad(): # Disable gradient tracking to save processing memory
    for images, targets in test_loader:
        logits = mlp_model(images)
        # torch.max extracts the index position containing the highest value (our predicted digit)
        _, predictions = torch.max(logits.data, dim=1)
        total_samples += targets.size(0)
        correct_predictions += (predictions == targets).sum().item()

final_accuracy = (correct_predictions / total_samples) * 100
print(f"Final Validated Model Accuracy: {final_accuracy:.2f}%")
```
